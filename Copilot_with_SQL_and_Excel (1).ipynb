Top 5 cities by revenue


```
Prompt for GitHub Copilot:

Find the total sales revenue for each city from the SalesData table.
Then display the top 5 cities with the highest revenue.

-- Total Sales Per City
SELECT
    City,
    SUM(PurchaseAmount) AS TotalSales
FROM SalesData
GROUP BY City
ORDER BY TotalSales DESC;

-- Top 5 Cities by Revenue
SELECT TOP 5
    City,
    SUM(PurchaseAmount) AS TotalRevenue
FROM SalesData
GROUP BY City
ORDER BY TotalRevenue DESC;

```

Find customers with purchases above average

```
Prompt for GitHub Copilot:

Find all customers whose PurchaseAmount is greater than the average PurchaseAmount in the SalesData table.

SELECT
    CustomerID,
    Name,
    City,
    PurchaseAmount
FROM SalesData
WHERE PurchaseAmount >
(
    SELECT AVG(PurchaseAmount)
    FROM SalesData
)
ORDER BY PurchaseAmount DESC;
```

format table)



```
Prompt for Excel Copilot:
Generate VBA code to clean the dataset by:
1. Removing duplicate rows.
2. Filling blank cells with appropriate values.
3. Formatting the data as an Excel Table.
4. Auto-fitting column widths.
5. Applying header formatting.


Sub CleanDataset()

    Dim ws As Worksheet
    Dim tbl As ListObject
    Dim lastRow As Long, lastCol As Long
    
    ' Set worksheet
    Set ws = ThisWorkbook.Sheets("Sheet1")
    
    ' Find last row and column
    lastRow = ws.Cells(ws.Rows.Count, 1).End(xlUp).Row
    lastCol = ws.Cells(1, ws.Columns.Count).End(xlToLeft).Column
    
    ' Remove duplicates (based on all columns)
    ws.Range(ws.Cells(1, 1), ws.Cells(lastRow, lastCol)).RemoveDuplicates Columns:=Array(1, 2, 3, 4, 5, 6), Header:=xlYes
    
    ' Fill blank cells with appropriate values
    Dim rng As Range
    Set rng = ws.Range(ws.Cells(2, 1), ws.Cells(lastRow, lastCol))
    
    Dim cell As Range
    For Each cell In rng
        If IsEmpty(cell.Value) Or cell.Value = "NA" Then
            If IsNumeric(cell.Offset(0, 3).Value) Then
                ' For numeric columns (Sales, Quantity)
                If cell.Column = 4 Or cell.Column = 5 Then
                    cell.Value = WorksheetFunction.Average(ws.Columns(cell.Column))
                End If
            Else
                ' For text columns (Customer_Name, Region, Product)
                If cell.Column = 1 Then cell.Value = "Sneha" ' Mode for Customer_Name
                If cell.Column = 2 Then cell.Value = "West"  ' Mode for Region
                If cell.Column = 3 Then cell.Value = "Monitor" ' Mode for Product
            End If
        End If
    Next cell
    
    ' Format as Excel Table
    On Error Resume Next
    Set tbl = ws.ListObjects("SalesTable")
    On Error GoTo 0
    
    If tbl Is Nothing Then
        Set tbl = ws.ListObjects.Add(xlSrcRange, ws.Range(ws.Cells(1, 1), ws.Cells(lastRow, lastCol)), , xlYes)
        tbl.Name = "SalesTable"
    End If
    
    ' Auto-fit column widths
    ws.Columns.AutoFit
    
    ' Apply header formatting
    With ws.Rows(1).Font
        .Bold = True
        .Color = vbWhite
    End With
    ws.Rows(1).Interior.Color = RGB(0, 102, 204) ' Blue header background
    
    MsgBox "Dataset cleaned and formatted successfully!", vbInformation

End Sub

```
